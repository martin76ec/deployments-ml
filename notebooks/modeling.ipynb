```python
import pandas as pd
```

```python
data_paths = {
  "movies": "dataset/movies.csv",
  "finantial": "dataset/finantials.csv",
  "opening": "dataset/opening_gross.csv"
}
```

```python
data = {
  "movies": pd.read_csv(data_paths["movies"]),
  "finantial": pd.read_csv(data_paths["finantial"]),
  "opening": pd.read_csv(data_paths["opening"]),
}
```

```python
numeric_columns_mask = (data["movies"].dtypes == float) | (data["movies"].dtypes == int)
numeric_columns = [column for column in numeric_columns_mask.index if numeric_columns_mask[column]]
movie_data = data["movies"][numeric_columns + ["movie_title"]]
```

```python
finantial_data = [["movie_title", "production_widget", "worldwide_gross"]]
```

```python
finantial_movie_data = pd.merge(finantial_data, movie_data, on = "movie_title", how = "left")
```
